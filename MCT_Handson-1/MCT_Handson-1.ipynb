Above code just gave us the column name is quan or qual but we want to know if qual=[column names] same for quan 
how we will attain this?

```python
#first use empty variable to store our values
quan=[]
qual=[]
for column_names in dataset.columns:
    if dataset[column_names].dtypes=="object":
        qual.append(column_names)
        #print("qual")
    else:
        quan.append(column_names)
        #print("quan")
```

Now this logic works fine so, we have to convert it into class so that we can use it multiple time

```python
def quan_qual():
    quan=[]
    qual=[]
    for column_names in dataset.columns:
        if dataset[column_names].dtypes=="object":
            qual.append(column_names)
            
        else:
            quan.append(column_names)
            
    return quan,qual
           
    
```

```python
quan_qual() #for future use we have to pass dataset into the function
```

```python
def quan_qual(dataset):
    quan=[]
    qual=[]
    for column_names in dataset.columns:
        if dataset[column_names].dtypes=="object":
            qual.append(column_names)
            
        else:
            quan.append(column_names)
            
    return quan,qual
```

```python
quan_qual(dataset)
```
