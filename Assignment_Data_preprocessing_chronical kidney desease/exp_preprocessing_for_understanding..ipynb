```python
import pandas as pd
```

```python
dataset=pd.read_csv("kidney_disease.csv")
```

```python
dataset.isna().sum() #we have nan values for sg column = 47
#our task is replace this sg value with "0" and rest of columns with median values.
```

```python
from Univariate import Univariate #Import class from .py file 
#From file name import class name 
```

```python
#from class name.function name()
quan,qual=Univariate.quan_qual(dataset)
```

```python
dataset.dtypes
```

```python
#If we use imputer all values will be replaced with meadian of dataset.
#but we want sg replaced by "0" so,
import numpy as np #(Create imputer model for all quan columns)
from sklearn.impute import SimpleImputer
imp_sg=SimpleImputer(strategy='constant',fill_value=0) 
#replace the nan values.
fitted_sg=imp_sg.fit_transform(dataset[["sg"]])
```

```python
dataset["sg"]=fitted_sg
```

```python
dataset
```

```python
dataset.isna().sum() #we can see now sg values are replaced by "0"
```

```python
import numpy as np #(Create imputer model for all quan columns)
from sklearn.impute import SimpleImputer
imp=SimpleImputer(missing_values=np.nan,strategy='median')#since we have lots of outliers we use median to 
#replace the nan values.
fitted_dataset1=imp.fit_transform(dataset[quan])
```

```python
fitted_dataset1
```

```python
df=pd.DataFrame(fitted_dataset1,columns=quan)
```

```python
pd.
```

```python
df.isna().sum()
```

```python
#good we have successfully replaced "sg" with value "0" and rest of quan columns with value of median
```

```python
#now we have to replace qual columns' nan values with strategy='most_frequent'
imp=SimpleImputer(missing_values=np.nan,strategy='most_frequent')
fitted_dataset2=imp.fit_transform(dataset[qual])
```

```python
df2=pd.DataFrame(fitted_dataset2,columns=qual)
```

```python
df2.isna().sum()
```

```python
#now merge df1 and df2 to make complete preproccessed dataframe.
result = pd.concat([df, df2], axis=1)
```
