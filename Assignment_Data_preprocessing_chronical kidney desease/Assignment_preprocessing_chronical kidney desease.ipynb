```python
lesser_de,greater_de=Univariate.detect_lesser_greater(quan,descriptive_As)
```

```python
#so in this dataset we have lots of outliers 
#we have to replace it for our dataset validation.
outliers_free=Univariate.Replace_outliers(dataset,lesser_de,greater_de,descriptive_As)
```

```python
outliers_free
```

```python
import numpy as np #(Create imputer model for all quan columns)
from sklearn.impute import SimpleImputer
imp=SimpleImputer(missing_values=np.nan,strategy='median')#since we have lots of outliers we use median to replace the nan values.
sim_imp=imp.fit_transform(dataset[quan])
```

```python
sim_imp
```

```python
#now convert it as a dataframe using pandas 
df=pd.DataFrame(sim_imp,columns=quan)
```

```python
df
```
