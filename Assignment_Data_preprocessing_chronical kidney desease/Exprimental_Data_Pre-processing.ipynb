```python
#Method 3: Replace according to dataset.(This method is suitable for this dataset)
dataset["salary"].fillna(0,inplace=True)
```

```python
#Method 4: consider this dataset as a semi-suppervised and find model to predict missing values.very rare method.
```

The above methods are suitable for one column 
what if Nan values present in more than 2 columns.

```python
def quan_qual(dataset):
    quan=[]
    qual=[]
    for column_names in dataset.columns:
        if dataset[column_names].dtypes=="object":
            qual.append(column_names)
            
        else:
            quan.append(column_names)
            
    return quan,qual
```

```python
quan,qual=quan_qual(dataset)
```

```python
quan
```

```python
dataset.isna().sum()
```

```python
#To sove the above challenge we have build in function.
import numpy as np
from sklearn.impute import SimpleImputer
imp = SimpleImputer(missing_values=np.nan, strategy='mean')
sim_imp=imp.fit_transform(dataset[quan])           #since this file can't process categorical data we pass quan
#imp.transform(dataset[quan])
```

```python
#dataset_Tab=pd.DataFrame(index=dataset[quan].index,columns=dataset[quan].columns) 
#if we run above code it gives dataframe with nan values)
dataset_Tab=pd.DataFrame(columns=dataset[quan].columns) #(this line of code gives only column names)
```
